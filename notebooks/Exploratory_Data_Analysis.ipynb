# Phân Tích Khám Phá Dữ Liệu (EDA) - Kinh Tế Việt Nam
## Dự án: Chuyển dịch cơ cấu kinh tế và Đóng góp của TFP
---
Notebook này thực hiện trực quan hóa chi tiết các bộ dữ liệu về GDP, Cơ cấu ngành và TFP.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['axes.titlesize'] = 16
plt.rcParams['axes.labelsize'] = 12
print("Setup complete")
```

```python
df_gdp_usage = pd.read_csv('../data/processed/gdp_usage_processed.csv', index_col=0)
df_gdp_sectors = pd.read_csv('../data/processed/gdp_sectors_processed.csv', index_col=0)
df_tfp = pd.read_csv('../data/processed/tfp_contribution_processed.csv', index_col=0)

print("Các cột trong GDP Usage:", df_gdp_usage.columns.tolist())
print("Dữ liệu đã sẵn sàng!")
```

```python
df_percent = df_gdp_sectors.copy()
cols_val = ['Giá trị: Nông lâm thủy sản', 'Giá trị: Công nghiệp xây dựng', 'Giá trị: Dịch vụ']
total = df_percent[cols_val].sum(axis=1)
for col in cols_val: df_percent[col + ' (%)'] = (df_percent[col] / total) * 100
pct_cols = [c + ' (%)' for c in cols_val]

plt.figure(figsize=(14, 7))
plt.stackplot(df_percent.index, df_percent[pct_cols[0]], df_percent[pct_cols[1]], df_percent[pct_cols[2]], 
              labels=['Nông nghiệp', 'Công nghiệp', 'Dịch vụ'], colors=['#2ecc71', '#e74c3c', '#3498db'], alpha=0.8)
plt.title('Cơ cấu Kinh tế Việt Nam (1986 - 2024)')
plt.legend(loc='upper right')
plt.show()
```

```python
plt.figure(figsize=(14, 7))
plt.plot(df_gdp_usage.index, df_gdp_usage['TieuDungCuoiCung'], label='Tiêu dùng', marker='o')
plt.plot(df_gdp_usage.index, df_gdp_usage['TongTaiSanCoDinh'], label='Đầu tư (Tài sản cố định)', marker='s')
plt.plot(df_gdp_usage.index, df_gdp_usage['ChanLechXuatNhapKhau'], label='Xuất khẩu ròng', marker='^')
plt.yscale('log')
plt.title('Thành phần GDP (Thang Log)')
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(2, 2, figsize=(20, 15))
plt.subplots_adjust(hspace=0.3, wspace=0.2)

# GDP Growth
axes[0,0].fill_between(df_gdp_usage.index, df_gdp_usage['GDP_TongSo'], color='skyblue', alpha=0.4)
axes[0,0].set_title('QUY MÔ GDP (Tỷ đồng)')

# Structural Shift
axes[0,1].stackplot(df_percent.index, df_percent[pct_cols[0]], df_percent[pct_cols[1]], df_percent[pct_cols[2]], 
                   labels=['Nông nghiệp', 'Công nghiệp', 'Dịch vụ'], colors=['#2ecc71', '#e74c3c', '#3498db'], alpha=0.8)
axes[0,1].set_title('CƠ CẤU NGÀNH (%)')
axes[0,1].legend(loc='lower left')

# Consumption vs Investment
width = 0.35
axes[1,0].bar(df_gdp_usage.index - width/2, df_gdp_usage['TieuDungCuoiCung'], width, label='Tiêu dùng')
axes[1,0].bar(df_gdp_usage.index + width/2, df_gdp_usage['TongTaiSanCoDinh'], width, label='Đầu tư')
axes[1,0].set_title('TIÊU DÙNG VS ĐẦU TƯ')
axes[1,0].legend()

# TFP
df_tfp[['Von_pct', 'LaoDong_pct', 'TFP_pct']].plot(kind='bar', stacked=True, ax=axes[1,1], color=['#95a5a6', '#f1c40f', '#e67e22'])
axes[1,1].set_title('ĐÓNG GÓP TFP (%)')
axes[1,1].legend(['Vốn', 'Lao động', 'TFP'], bbox_to_anchor=(1, 1))

plt.suptitle('DASHBOARD TỔNG QUAN KINH TẾ VIỆT NAM', fontsize=24)
plt.show()
```
